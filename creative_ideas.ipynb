# Creative Idea Generator

Enter an everyday object and get an unusual, practical use from an OpenAI model.

**Setup:** Run the install cell, then run the generator. Your API key is requested securely at runtime and is never stored in this notebook.

```python
!pip -q install openai
```

```python
from getpass import getpass
from openai import OpenAI

client = OpenAI(api_key=getpass("OpenAI API key: "))

def suggest_idea(item):
    response = client.responses.create(
        model="gpt-6-luna",
        input=f"Suggest ONE unusual but practical use for this everyday object: {item}. Keep it concise."
    )
    return response.output_text

item = input("Enter an everyday object: ").strip()
if item:
    print(suggest_idea(item))
else:
    print("Please enter an object and run this cell again.")
```
